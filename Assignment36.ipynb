# HuggingFace with Langchain
## 1. Integrate the HuggingFace model using Langchain Wrappers.
## 2. Replace OpenAI with HuggingFace in an LLM chain.
## 3. Test with multiple prompts.

```python
!pip install -U langchain-huggingface huggingface_hub
```

```python
from dotenv import load_dotenv
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from langchain_core.prompts import ChatPromptTemplate
import os

# Load .env
load_dotenv()

# Get Hugging Face token
token = os.getenv("HF_TOKEN")
```

```python

# Hugging Face model
llm = HuggingFaceEndpoint(
    repo_id="meta-llama/Llama-3.1-8B-Instruct",
    huggingfacehub_api_token=token,
    max_new_tokens=200,
    temperature=0.7
)
```

```python

# Convert HuggingFace endpoint into LangChain chat model
chat_model = ChatHuggingFace(llm=llm)

# LangChain prompt
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI assistant."),
    ("human", "{question}")
])
```

```python

# Create LangChain chain
chain = prompt | chat_model

# Multiple prompts
questions = [
    "What is Generative AI?",
    "Explain Machine Learning in simple words.",
    "What is Python used for?",
    "What is RAG in Generative AI?"
]
```

```python

# Test the chain
for question in questions:
    response = chain.invoke({
        "question": question
    })
    print("QUESTION:", question)
    print("ANSWER:", response.content)
```

# Chat Prompt Template with HuggingFace
## 1. Create a ChatPromptTemplate.
## 2. Use system + human messages.
## 3. Generate responses using HuggingFace-backed LLM.
## 4. Compare with normal prompt template.

```python
from dotenv import load_dotenv
import os

from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace
from langchain_core.prompts import ChatPromptTemplate, PromptTemplate

load_dotenv()

token = os.getenv("HF_TOKEN")

# Hugging Face LLM
llm = HuggingFaceEndpoint(
    repo_id="meta-llama/Llama-3.1-8B-Instruct",
    huggingfacehub_api_token=token,
    max_new_tokens=200,
    temperature=0.7
)

# Convert to Chat Model
chat_model = ChatHuggingFace(llm=llm)


# 1. ChatPromptTemplate

chat_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI teacher."),
    ("human", "{question}")
])

# Create chain
chain = chat_prompt | chat_model

# Generate response
response = chain.invoke({
    "question": "Explain Generative AI in simple words."
})

print(response.content)

# 2. Normal PromptTemplate

normal_prompt = PromptTemplate(
    template="Explain {topic} in simple words.",
    input_variables=["topic"]
)

normal_chain = normal_prompt | llm

response2 = normal_chain.invoke({
    "topic": "Generative AI"
})

print("Normal PromptTemplate Response:",response2)
```
